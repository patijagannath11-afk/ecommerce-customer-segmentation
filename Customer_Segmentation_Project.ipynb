```python
import pandas as pd
from sqlalchemy import create_engine

# --- CREATE THE *NEW* CONNECTION ---
# We are using "mysql+pymysql" in the string now

db_string = "mysql+pymysql://root:7008940177@localhost:3306/jagannath2"

# Create the engine
try:
    engine = create_engine(db_string)
    print("Connection Successful! (PyMySQL worked!)")
except Exception as e:
    print(f"Connection Failed: {e}")
```

```python
# --- 1. Write the SQL Query ---
# This is the query to get all the data we need for RFM
# We JOIN the tables inside the database *before* pulling the data

sql_query = """
SELECT 
    c.customer_unique_id,  -- The ID for the person
    o.order_purchase_timestamp, -- When they bought
    op.payment_value         -- How much they paid
FROM `order` o
JOIN customers c ON o.customer_id = c.customer_id
JOIN order_payments op ON o.order_id = op.order_id
WHERE o.order_status = 'delivered';"""

# --- 2. Run the Query with Pandas ---
# This tells pandas to run our query using our engine
# and store the results in a "DataFrame" (a table) called 'df'
try:
    print("Running SQL query... This might take a moment...")
    df = pd.read_sql_query(sql_query, con=engine)
    
    print("Data imported successfully!")
    # Show the first 5 rows
    print(df.head())
    
    # Show how many rows we got
    print(f"\nTotal rows imported: {len(df)}")
    
except Exception as e:
    print(f"Query Failed: {e}")
```
