```python
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df = pd.read_csv("GroceryStoreDataSet.csv", names = ["products"], sep = ",")
df.head()
```

```python
x=df["products"].apply(lambda x:x.split(","))
```

```python
data = list(x)
data
```

```python
# lets transdorm the list , with one hot encoding
from mlxtend.preprocessing import TransactionEncoder
a = TransactionEncoder()
```

```python
a_data = a.fit(data).transform(data)
df1 = pd.DataFrame(a_data,columns=a.columns_)
df1
```

```python
# set a threshold for the value for support and then calu=culate the support value
df2 = apriori(df1, min_support = 0.15, use_colnames = True , verbose = 1)
df2
```
